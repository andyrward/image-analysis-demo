# Image Analysis Demo: Synthetic Spot Generation

This notebook demonstrates how to generate synthetic images with spots at sub-pixel coordinates for testing image analysis algorithms.

## Overview

The `generate_spots_image()` function creates test images with NxN spots placed on a grid with random sub-pixel offsets. This is useful for:
- Testing spot detection algorithms
- Validating image analysis pipelines
- Benchmarking sub-pixel localization accuracy
- Simulating realistic imaging conditions with noise

## Setup

Import required libraries and the spot generation function.

```python
import numpy as np
import matplotlib.pyplot as plt
from main import generate_spots_image

# Set random seed for reproducibility
np.random.seed(42)

# Configure matplotlib for better visualization
plt.rcParams['figure.figsize'] = (12, 10)
plt.rcParams['image.cmap'] = 'viridis'
```

## Example 1: Basic Spot Generation

Generate a simple 5x5 grid of spots with moderate noise.

```python
# Parameters for basic example
n = 5                    # 5x5 grid (25 spots total)
spot_width = 3.0         # Gaussian sigma in pixels
signal = 100.0           # Peak intensity of spots
noise_magnitude = 5.0    # Standard deviation of noise
image_size = 512         # Image size in pixels

# Generate the image
image = generate_spots_image(n, spot_width, signal, noise_magnitude, image_size)

# Display the image
plt.figure(figsize=(10, 10))
plt.imshow(image, cmap='hot')
plt.colorbar(label='Intensity')
plt.title(f'{n}x{n} Spots with Sub-pixel Offsets\nSpot Width={spot_width}px, Signal={signal}, Noise={noise_magnitude}')
plt.xlabel('X (pixels)')
plt.ylabel('Y (pixels)')
plt.tight_layout()
plt.show()

print(f"Image shape: {image.shape}")
print(f"Value range: [{image.min():.2f}, {image.max():.2f}]")
print(f"Mean: {image.mean():.2f}")
print(f"Standard deviation: {image.std():.2f}")
```

## Example 2: Comparing Different Noise Levels

Demonstrate how noise affects the image quality.

```python
# Generate images with different noise levels
noise_levels = [0, 2, 5, 10, 20]

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for idx, noise in enumerate(noise_levels):
    image = generate_spots_image(n=4, spot_width=3.0, signal=100.0, 
                                 noise_magnitude=noise, image_size=256)
    
    axes[idx].imshow(image, cmap='hot', vmin=0, vmax=120)
    axes[idx].set_title(f'Noise σ = {noise}')
    axes[idx].set_xlabel('X (pixels)')
    axes[idx].set_ylabel('Y (pixels)')

# Hide the last unused subplot
axes[-1].axis('off')

plt.suptitle('Effect of Noise on Spot Images (4x4 grid)', fontsize=14, y=1.00)
plt.tight_layout()
plt.show()
```

## Example 3: Different Spot Widths

Explore how spot width (Gaussian sigma) affects the appearance.

```python
# Generate images with different spot widths
spot_widths = [1.5, 2.5, 4.0, 6.0]

fig, axes = plt.subplots(2, 2, figsize=(12, 12))
axes = axes.flatten()

for idx, width in enumerate(spot_widths):
    image = generate_spots_image(n=4, spot_width=width, signal=100.0, 
                                 noise_magnitude=3.0, image_size=256)
    
    axes[idx].imshow(image, cmap='hot')
    axes[idx].set_title(f'Spot Width σ = {width} pixels')
    axes[idx].set_xlabel('X (pixels)')
    axes[idx].set_ylabel('Y (pixels)')

plt.suptitle('Effect of Spot Width on Image Appearance (4x4 grid)', fontsize=14, y=1.00)
plt.tight_layout()
plt.show()
```

## Example 4: Dense vs Sparse Grids

Compare different grid densities.

```python
# Generate images with different grid sizes
grid_sizes = [3, 5, 8, 12]

fig, axes = plt.subplots(2, 2, figsize=(12, 12))
axes = axes.flatten()

for idx, n in enumerate(grid_sizes):
    image = generate_spots_image(n=n, spot_width=2.5, signal=100.0, 
                                 noise_magnitude=4.0, image_size=512)
    
    axes[idx].imshow(image, cmap='hot')
    axes[idx].set_title(f'{n}x{n} Grid ({n*n} spots total)')
    axes[idx].set_xlabel('X (pixels)')
    axes[idx].set_ylabel('Y (pixels)')

plt.suptitle('Different Grid Densities', fontsize=14, y=1.00)
plt.tight_layout()
plt.show()
```

## Example 5: Analyzing a Single Spot Profile

Zoom in on a single spot to see its Gaussian profile.

```python
# Generate a simple image with larger spots
image = generate_spots_image(n=3, spot_width=5.0, signal=100.0, 
                             noise_magnitude=2.0, image_size=256)

# Extract a region around the center spot
center = 256 // 2
region_size = 60
spot_region = image[center-region_size:center+region_size, 
                    center-region_size:center+region_size]

# Create visualizations
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 2D view
im = axes[0].imshow(spot_region, cmap='hot')
axes[0].set_title('2D View of Single Spot')
axes[0].set_xlabel('X (pixels)')
axes[0].set_ylabel('Y (pixels)')
plt.colorbar(im, ax=axes[0], label='Intensity')

# Cross-section along X
y_center = spot_region.shape[0] // 2
axes[1].plot(spot_region[y_center, :], 'r-', linewidth=2)
axes[1].set_title('Horizontal Cross-Section')
axes[1].set_xlabel('X (pixels)')
axes[1].set_ylabel('Intensity')
axes[1].grid(True, alpha=0.3)

# Cross-section along Y
x_center = spot_region.shape[1] // 2
axes[2].plot(spot_region[:, x_center], 'b-', linewidth=2)
axes[2].set_title('Vertical Cross-Section')
axes[2].set_xlabel('Y (pixels)')
axes[2].set_ylabel('Intensity')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Example 6: Interactive Parameter Exploration

Try different parameter combinations to see their effects.

```python
# Define your custom parameters here
custom_n = 6                    # Grid size (try 3-15)
custom_spot_width = 3.5         # Spot width in pixels (try 1-10)
custom_signal = 120.0           # Signal intensity (try 50-200)
custom_noise = 8.0              # Noise level (try 0-20)
custom_image_size = 512         # Image size (try 256, 512, 1024)

# Generate the custom image
custom_image = generate_spots_image(
    n=custom_n,
    spot_width=custom_spot_width,
    signal=custom_signal,
    noise_magnitude=custom_noise,
    image_size=custom_image_size
)

# Display results
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Image display
im = axes[0].imshow(custom_image, cmap='hot')
axes[0].set_title(f'Custom Image: {custom_n}x{custom_n} Spots')
axes[0].set_xlabel('X (pixels)')
axes[0].set_ylabel('Y (pixels)')
plt.colorbar(im, ax=axes[0], label='Intensity')

# Histogram
axes[1].hist(custom_image.flatten(), bins=100, alpha=0.7, color='blue', edgecolor='black')
axes[1].set_title('Intensity Histogram')
axes[1].set_xlabel('Intensity Value')
axes[1].set_ylabel('Frequency')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Print statistics
print("\nImage Statistics:")
print(f"  Shape: {custom_image.shape}")
print(f"  Total spots: {custom_n * custom_n}")
print(f"  Min value: {custom_image.min():.2f}")
print(f"  Max value: {custom_image.max():.2f}")
print(f"  Mean: {custom_image.mean():.2f}")
print(f"  Std dev: {custom_image.std():.2f}")
print(f"  Signal-to-Noise Ratio (approx): {custom_signal / custom_noise:.2f}" if custom_noise > 0 else "  SNR: Infinite (no noise)")
```

## Example 7: Saving Images for Further Analysis

Generate and save images for use in other analysis tools.

```python
from PIL import Image

# Generate a test image
test_image = generate_spots_image(n=5, spot_width=3.0, signal=100.0, 
                                  noise_magnitude=5.0, image_size=512)

# Normalize to 0-255 range for saving
if test_image.max() > test_image.min():
    image_normalized = (test_image - test_image.min()) / (test_image.max() - test_image.min()) * 255
else:
    image_normalized = np.full_like(test_image, 128)
image_normalized = image_normalized.astype(np.uint8)

# Save as PNG
img = Image.fromarray(image_normalized)
img.save('generated_spots.png')

print("✓ Image saved as 'generated_spots.png'")
print(f"  Dimensions: {image_normalized.shape}")
print(f"  Data type: {image_normalized.dtype}")

# Also save the raw data as numpy array for precise analysis
np.save('generated_spots_raw.npy', test_image)
print("\n✓ Raw data saved as 'generated_spots_raw.npy'")
print("  (This preserves exact floating-point values for analysis)")
```

3. **Noise simulation**: Additive Gaussian noise simulates realistic imaging conditions
